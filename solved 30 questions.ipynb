```python
29
import pandas as pd

sales = pd.Series(
    [600, 450, 750, 300],
    index=['Laptop', 'Mouse', 'Phone', 'Keyboard']
)

print(sales)

# Select one label
print(sales['Phone'])

# Filter sales above 500
print(sales[sales > 500])

# Check whether a label exists
print('Laptop' in sales.index)

# Convert Series to dictionary
sales_dict = sales.to_dict()
print(sales_dict)
```
